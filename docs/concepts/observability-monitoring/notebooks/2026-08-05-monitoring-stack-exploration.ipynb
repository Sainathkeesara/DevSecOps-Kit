# Monitoring stack exploration

I started this notebook to get a feel for how Grafana and Prometheus fit together. I'm following along with the basic setup and writing down what tripped me up.

```python
# Minimal Prometheus config to scrape a local target
prometheus_config = """
global:
  scrape_interval: 15s

scrape_configs:
  - job_name: 'prometheus'
    static_configs:
      - targets: ['localhost:9090']
"""

print(prometheus_config)
```

## What I learned

The first thing that tripped me up was the indentation in the YAML. Prometheus rejected the config with a generic parse error that pointed at the wrong line. Once I switched my editor to insert spaces instead of tabs, the config loaded cleanly.

```python
# Simple Grafana dashboard JSON for a single target
dashboard = {
    "title": "Local Prometheus Target",
    "panels": [
        {
            "title": "Scrape Duration",
            "type": "graph",
            "datasource": "Prometheus",
            "targets": [
                {
                    "expr": "rate(prometheus_target_scrape_pool_targets[1m])",
                    "legendFormat": "{{job}}"
                }
            ]
        }
    ]
}

import json
print(json.dumps(dashboard, indent=2))
```
